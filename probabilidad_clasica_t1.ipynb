# Ejercicio 1
Al lanzar dos dados, calcula la probabilidad de que la suma de sus caras sea exactamente 5.

El espacio muestral al lanzar dos dados consta de 6x6=36 resultados posibles. Los casos favorables en los que la suma de las caras es exactamente 5 son cuatro: (1, 4), (2, 3), (3, 2) y (4, 1).
La probabilidad es: $$P = \frac{4}{36} = \frac{1}{9}$$

```python
casos_totales = 6 * 6
casos_favorables = sum(1 for dado1 in range(1, 7) for dado2 in range(1, 7) if dado1 + dado2 == 5)
probabilidad = casos_favorables / casos_totales
print(casos_favorables)
print(f"Probabilidad: {probabilidad} ({casos_favorables}/{casos_totales})")
```

```python
from itertools import product

# S: Todas las combinaciones posibles al lanzar 2 dados de 6 caras
S = list(product(range(1, 7), repeat=2))

# A: Solo aquellas tuplas cuya suma sea exactamente 5
A = [x for x in S if sum(x) == 5]

print("Ejercicio 1")
print("Resultados posibles:", len(S))
print("Casos favorables:", len(A)) # Imprime 4
print("Probabilidad:", len(A) / len(S))
```

# Ejercicio 2
En una contraseña de 3 dígitos, del 0 al 9, calcula la probabilidad de que ninguno de los dígitos se repita.

Una contraseña de 3 dígitos (usando números del o al 9) tiene 10x10x10 = 1000 combinaciones totales posibles.
Para que ningún número se repita, el primer dígito tiene 10 opciones disponibles, el segundo tiene 9 opciones (excluyendo el primer número), el tercer dígito tiene 8 opciones (excluyendo lo dos anteriores) por lo cual los casos favorables son 10x9x8=720.
La probabilidad es:

$$P = \frac{720}{1000} = \frac{18}{25}=0.72$$

```python
from itertools import product

S = list(product(range(10), repeat=3))

A=[x for x in S if len(set(x)) == 3]

print("Ejercicio 2")
print(f"Resultados posibles: {len(S)}")
print(f"Casos favorables: {len(A)}")
print(f"Probabilidad: {len(A) / len(S)}")

```

# Ejercicio 3
Tienes 8 canciones distintas y las pones en un orden aleatorio, con todos los órdenes igualmente probables. Calcula la probabilidad de que tu favorita quede primera y la que menos te gusta quede última.

El número total de formas de ordenar 8 canciones distintas es 8! (40,320). Si fijamos la canción favorita en la primera posición y la que menos nos gusta en la última, quedan 6 canciones para organizar en las posiciones intermedias. Estas 6 canciones se pueden ordenar de 6! formas (720).
La probabilidad es:
$$P = \frac{6!}{8!} = \frac{6!}{(8)(7)(6!)} = \frac{1}{56}$$

```python
from itertools import permutations

# S: Todas las permutaciones posibles de las 8 canciones (representadas del 1 al 8)
S = list(permutations(range(1, 9)))

# A: La canción favorita (ej. la 1) en el primer índice [0], y la que menos te gusta (ej. la 8) en el último [-1]
A = [x for x in S if x[0] == 1 and x[-1] == 8]

print("Ejercicio 3")
print("Resultados posibles:", len(S))
print("Casos favorables:", len(A))
print("Probabilidad:", len(A) / len(S))
```

# Ejercicio 5
Ana, Bruno, Carla y Diego hacen un amigo secreto: cada quien saca un regalo al azar y todos los repartos son igualmente probables. Calcula la probabilidad de que nadie saque su propio regalo.



```python
import math
n = 4
casos_totales = math.factorial(n)
# Cálculo de desarreglos usando la fórmula de inclusión-exclusión
desarreglos = sum((-1)**k * (math.factorial(n) // math.factorial(k)) for k in range(n + 1))
probabilidad = desarreglos / casos_totales

print(f"Probabilidad: {probabilidad} ({desarreglos}/{casos_totales})")
```

```python
from itertools import permutations

# S: Reparto de 4 regalos (1 al 4) entre 4 personas en posiciones (0 al 3)
S = list(permutations(range(1, 5)))

# A: Se usa la función all() para asegurar que NADIE en el índice 'i' tenga el regalo 'i+1'
A = [x for x in S if all(x[i] != i + 1 for i in range(4))]

print("Ejercicio 5")
print("Resultados posibles:", len(S))
print("Casos favorables (desarreglos):", len(A)) # Imprime 9
print("Probabilidad:", len(A) / len(S))
```
